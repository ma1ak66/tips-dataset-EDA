# Exploratory Data Analysis (EDA) – Tips Dataset

## Project Overview
This project performs an exploratory data analysis (EDA) on the Seaborn **Tips** dataset. The analysis focuses on understanding tipping behavior and exploring relationships between the bill amount, tip, party size, day, time, and smoker status.

### Objectives
- Inspect the dataset structure and data quality.
- Explore the relationship between total bill and tip.
- Compare average tips across days.
- Compare tip distributions for smokers and non-smokers.
- Examine correlations among numerical variables.
- Summarize the main observations from the analysis.

**Tools:** Python, Pandas, Matplotlib, Seaborn

## 1. Setup & Data Loading

Import the required libraries and load the built-in Seaborn `tips` dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
tips = sns.load_dataset("tips")

tips.head()
```

## 2. Dataset Overview & Data Quality

Before visual analysis, inspect the shape, columns, data types, missing values, and duplicate rows.

```python
print(f"Rows: {tips.shape[0]}")
print(f"Columns: {tips.shape[1]}")

display(tips.info())

print("\nMissing values:")
display(tips.isnull().sum())

print(f"Duplicate rows: {tips.duplicated().sum()}")
```

```python
# Descriptive statistics for numerical variables
tips.describe().round(2)
```

## 3. Scatter Plot: Total Bill vs. Tip

**Goal:** Analyze the relationship between the total bill amount and the tip, while distinguishing lunch and dinner observations.

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="time")
plt.title("Total Bill vs. Tip by Time of Day")
plt.xlabel("Total Bill ($)")
plt.ylabel("Tip ($)")
plt.tight_layout()
plt.savefig("scatter_total_bill_vs_tip.png", dpi=300, bbox_inches="tight")
plt.show()
```

### Observation
The scatter plot shows a generally positive relationship: larger total bills tend to be associated with larger tips. The plot also shows observations from both lunch and dinner.

## 4. Bar Plot: Average Tip by Day

**Goal:** Compare the average tip amount across the different days represented in the dataset.

```python
avg_tip_by_day = tips.groupby("day", observed=False)["tip"].mean().sort_values(ascending=False)
display(avg_tip_by_day.round(2))

plt.figure(figsize=(8, 5))
sns.barplot(data=tips, x="day", y="tip", errorbar=None)
plt.title("Average Tip by Day of the Week")
plt.xlabel("Day of the Week")
plt.ylabel("Average Tip ($)")
plt.tight_layout()
plt.show()
```

### Observation
Average tip amounts vary across the days in the dataset. The grouped summary above makes the differences directly comparable.

## 5. Box Plot: Tip Distribution for Smokers vs. Non-Smokers

**Goal:** Compare the distribution, median, spread, and possible outliers of tips for smokers and non-smokers.

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=tips, x="smoker", y="tip")
plt.title("Tip Distribution: Smoker vs. Non-Smoker")
plt.xlabel("Smoker Status")
plt.ylabel("Tip ($)")
plt.savefig("tip_distribution_smoker.png", dpi=300, bbox_inches="tight")
plt.tight_layout()
plt.show()
```

### Observation
The box plot allows comparison of the median, interquartile range, and potential outliers in tip amounts between the two smoker-status groups.

## 6. Correlation Heatmap

**Goal:** Examine pairwise correlations among the numerical variables: `total_bill`, `tip`, and `size`.

```python
corr_matrix = tips[["total_bill", "tip", "size"]].corr()
display(corr_matrix.round(2))

plt.figure(figsize=(6, 4))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap of Numerical Variables")
plt.tight_layout()
plt.savefig("correlation_heatmap.png", dpi=300, bbox_inches="tight")
plt.show()
```

### Observation
The correlation matrix provides a compact view of the linear relationships among bill amount, tip, and party size. In particular, it can be used to assess how strongly tip changes with the total bill.

## 7. Key Insights

Based on the analysis:
- The dataset contains **244 observations and 7 variables**.
- `total_bill` and `tip` show a positive relationship in the scatter plot.
- Average tip values differ across the days represented in the dataset.
- The smoker and non-smoker groups have different tip distributions, which can be compared using their medians and spread.
- The correlation heatmap summarizes the relationships among `total_bill`, `tip`, and `size`.

These observations are descriptive findings from this dataset and should not be interpreted as causal relationships.
